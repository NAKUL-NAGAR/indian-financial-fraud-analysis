```python
master = transactions.merge(customers,on="Customer_ID",how="left",suffixes=("", "_Customer")) # Merge transactions with customers
```

```python
master = master.merge(cards,on=["Card_ID", "Customer_ID"],how="left",suffixes=("", "_Card")) # Merge the resulting DataFrame with cards on Card_ID and Customer_ID
```

```python
master = master.merge(merchants,on="Merchant_ID",how="left",suffixes=("", "_Merchant")) # Merge the resulting DataFrame with merchants on Merchant_ID
```

```python
print(master.shape)  # Print the shape of the merged DataFrame
print(master.head())  # Print the first 5 rows of the merged DataFrame
```

```python
master["Fraud_Amount"] = np.where(master["Fraud_Flag"] == 1,master["Transaction_Amount"],0) # Create a new column for fraud amount
```

```python
master["Successful_Flag"] = np.where(master["Transaction_Status"] == "Successful",1,0) # Create a new column for successful transactions
```

```python
master["Declined_Flag"] = np.where(master["Transaction_Status"] == "Declined",1,0) # Create a new column for declined transactions
```

```python
fraud_count = master["Fraud_Flag"].sum() # Calculate the total number of fraudulent transactions

print(fraud_count) # Print the total number of fraudulent transactions
```

```python
fraud_rate = (master["Fraud_Flag"].mean()*100) # Calculate the fraud rate as a percentage of total transactions
print(round(fraud_rate, 2), "%") # Print the fraud rate as a percentage
```

```python
fraud_payment = pd.crosstab(master["Payment_Method"],master["Fraud_Flag"],normalize="index") * 100
print(fraud_payment.round(2))
```

```python
fraud_category = master.groupby("Merchant_Category").agg(Transactions=("Transaction_ID", "count"),Fraud_Count=("Fraud_Flag", "sum"),Transaction_Amount=("Transaction_Amount", "sum")) # Group by merchant category and calculate transaction metrics
fraud_category["Fraud_Rate"] = (fraud_category["Fraud_Count"]/ fraud_category["Transactions"]* 100) # Calculate fraud rate as a percentage
print(fraud_category.sort_values("Fraud_Rate",ascending=False)) # Print the sorted fraud category data
```

```python
fraud_segment = master.groupby("Customer_Segment").agg(Transactions=("Transaction_ID", "count"),Fraud_Count=("Fraud_Flag", "sum"),Amount=("Transaction_Amount", "sum")) # Group by customer segment and calculate transaction metrics
fraud_segment["Fraud_Rate"] = (fraud_segment["Fraud_Count"]/ fraud_segment["Transactions"] * 100) # Calculate fraud rate as a percentage
print(fraud_segment) # Print the fraud segment data
```

```python
fraud_state = master.groupby("Customer_State").agg(Transactions=("Transaction_ID", "count"),Fraud_Count=("Fraud_Flag", "sum"),Amount=("Transaction_Amount", "sum")) # Group by customer state and calculate transaction metrics
fraud_state["Fraud_Rate"] = (fraud_state["Fraud_Count"]/ fraud_state["Transactions"] * 100) # Calculate fraud rate as a percentage
print(fraud_state.sort_values("Fraud_Rate",ascending=False)) # Print the sorted fraud state data
```

```python
international = master.groupby("Is_International").agg(Transactions=("Transaction_ID", "count"),Fraud_Count=("Fraud_Flag", "sum"),Amount=("Transaction_Amount", "sum")) # Group by international transactions and calculate transaction metrics
international["Fraud_Rate"] = ( international["Fraud_Count"]/ international["Transactions"] * 100) # Calculate fraud rate as a percentage
print(international)  # Print the international transaction data
```

```python
#Fraud by card type

card_analysis = master.groupby("Card_Type").agg(Transactions=("Transaction_ID", "count"),Fraud_Count=("Fraud_Flag", "sum"),Amount=("Transaction_Amount", "sum")) # Group by card type and calculate transaction metrics
card_analysis["Fraud_Rate"] = (card_analysis["Fraud_Count"]/ card_analysis["Transactions"] * 100) # Calculate fraud rate as a percentage
print(card_analysis) # Print the card type analysis data
```

```python
#Fraud by card status

card_status_analysis = master.groupby("Card_Status").agg(Transactions=("Transaction_ID", "count"),Fraud_Count=("Fraud_Flag", "sum")) # Group by card status and calculate transaction metrics
card_status_analysis["Fraud_Rate"] = (card_status_analysis["Fraud_Count"]/ card_status_analysis["Transactions"] * 100) # Calculate fraud rate as a percentage
print(card_status_analysis) # Print the card status analysis data
```
